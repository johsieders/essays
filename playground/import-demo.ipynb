# Importing Variables from Python Programs

How to reuse computations across notebooks.

## Approach 1: Import Functions from math4phys

````python
from sympy import symbols, latex
from IPython.display import display, Math
from math4phys.diff_ops import gradient, divergence, laplacian

# Define variables
x, y, z = symbols('x y z')

# Define a scalar field
f = x**2 + y**2 + z**2

# Use imported functions
grad_f = gradient(f, [x, y, z])
lap_f = laplacian(f, [x, y, z])

display(Math(r"f = " + latex(f)))
display(Math(r"\nabla f = " + latex(grad_f)))
display(Math(r"\nabla^2 f = " + latex(lap_f)))
````

## Approach 2: Create Helper Functions

Put commonly used expressions in functions:

````python
from sympy import sqrt

def radial_distance(x, y, z):
    """Common radial distance function."""
    return sqrt(x**2 + y**2 + z**2)

r = radial_distance(x, y, z)
grad_r = gradient(r, [x, y, z])

display(Math(r"\text{Radial distance: } " + latex(r)))
display(Math(r"\nabla r = " + latex(grad_r)))
````

## Approach 3: Store Common Variables

Define standard variables once and reuse:

````python
# Gravitational potential (common in physics)
V_grav = -1 / radial_distance(x, y, z)
E_field = -gradient(V_grav, [x, y, z])

display(Math(r"V = " + latex(V_grav)))
display(Math(r"\vec{E} = -\nabla V = " + latex(E_field)))
````

## Summary

**Best practices:**

1. ✅ **Import functions from math4phys**
   ```python
   from math4phys.diff_ops import gradient, laplacian
   result = gradient(expr, vars)
   ```

2. ✅ **Create helper functions** for common patterns
   ```python
   def radial_distance(x, y, z):
       return sqrt(x**2 + y**2 + z**2)
   ```

3. ✅ **Store intermediate results** as variables
   ```python
   V = -1/r  # Reuse in multiple cells
   ```

**The pattern:** Import functions from `math4phys`, define helpers for your specific use case.
