# Hide Code Demo

This demonstrates hiding code cells while showing outputs.

```python
# This code cell is VISIBLE (no tags)
from sympy import symbols, latex
from IPython.display import display, Math

x = symbols('x')
f = x**2

display(Math(r"\text{Visible code: } f(x) = " + latex(f)))
```

```python
# This code cell is HIDDEN but toggleable (has 'hide-input' tag)
from math4phys.diff_ops import gradient

grad_f = gradient(f, [x])

display(Math(r"\text{Hidden code (toggleable): } \nabla f = " + latex(grad_f[0])))
```

```python
# This code cell is COMPLETELY REMOVED (has 'remove-input' tag)
from sympy import diff

second_deriv = diff(f, x, 2)

display(Math(r"\text{Removed code (no toggle): } \frac{d^2}{dx^2} f = " + latex(second_deriv)))
```

## Summary

- **No tag**: Code is visible
- **`hide-input` tag**: Code hidden with "Click to show" button
- **`remove-input` tag**: Code completely removed, no button
