# Accessing Python Variables in Notebooks

**Bottom line: Use `display(Math(...))` for rendered LaTeX in notebooks.**

Other methods don't render properly in Jupyter Book.

## Method 1: Direct Display with Math (✅ WORKS)

```python
from sympy import symbols, latex, diff, Matrix
from IPython.display import display, Math
from myst_nb import glue

# Create symbolic variables
x, y = symbols('x y')
f = x**2 + 2*x*y + y**2
grad_f = Matrix([diff(f, x), diff(f, y)])

# THIS WORKS - renders beautifully
display(Math(r"f(x,y) = " + latex(f)))
display(Math(r"\nabla f = " + latex(grad_f)))

f_value = f.subs({x: 1, y: 2})
display(Math(r"\text{At } (1,2): f(1,2) = " + str(f_value)))
```

## Method 2: Glue for Cross-File References

```python
# Glue LaTeX strings for OTHER files to use
glue("my_function", latex(f), display=False)
glue("my_gradient", latex(grad_f), display=False)
glue("my_value", int(f_value), display=False)

print("✓ Variables glued for cross-file reference")
print("\nTo reference from glue-import.md, see that file for syntax.")
```

## Example: Product Rule

```python
# Product rule example
g = x**3
fg = f * g
grad_fg = Matrix([diff(fg, x), diff(fg, y)])

display(Math(r"\text{Let } g(x) = " + latex(g)))
display(Math(r"\nabla(fg) = " + latex(grad_fg)))
```

## Summary

**What works in Jupyter Book:**
- ✅ `display(Math(latex(...)))` - renders beautifully
- ✅ `glue(...)` - for cross-file references (see glue-import.md)

**What doesn't work well:**
- ❌ `display(Markdown(...))` - shows as object, not rendered
- ❌ Direct variable access in markdown cells - impossible

**The pattern:** Generate LaTeX in code cells, display with `Math()`
